```python
import pandas as pd
```

```python
df = pd.read_csv('OnlineAnalysis.csv')
#print(df.head())
print(df.columns)
```

```python
perf_metrics = ['Data Source','MAXABSIMPXTE (m)', 'AVGABSIMPXTE (m)']
df2 = df[perf_metrics]
df2["AVG OF PEAKS"] = 0
df2["AVG OF ABS"] = 0
#print(df2.info())
print(df2.head())
```

```python
start_index = 2
num_of_samples = 3

end_index = start_index + num_of_samples   # El rango es 2 hasta 4. El 5 no se incluye
                                            #cuando llega a 5 sale de la funcion. Es como less or equal to 5

print(f"Rows from: {start_index} to {end_index - 1}") # Se pone el -1 solo para imprimir el rango correcto


peaks_col =  df2['MAXABSIMPXTE (m)'].iloc[start_index:end_index] #Del df2 toma la col MAXABSIMPXTE y ... 
                                                                # los renglones 2 hasta 4. Cuando llega a 5 break.


abs_col =  df2['AVGABSIMPXTE (m)'].iloc[start_index:end_index] #Del df2 toma la col AVGABSIMPXTE y ... 
                                                                # los renglones 2 hasta 4. Cuando llega a 5 break.

print(peaks_col) 
print(abs_col) 
```

```python
temp_mean_peaks = peaks_col.mean()
temp_mean_abs = abs_col.mean()

print(f"Results from index: {start_index} to {end_index - 1}")

print(f"Average of MAXES: {temp_mean_peaks}")
print(f"Average of ABS: {temp_mean_abs}")
```

```python
df2.iloc[end_index - 1,3] = temp_mean_peaks # [row = 4 ,  col = "MAXABSIMPXTE" ] = temp_mean_peaks
df2.iloc[end_index - 1, 4] = temp_mean_abs

print(df2.head())
```

```python
start_index = end_index
```

```python
def get_ranges(df,start_index,samples):


    #ejemplo 
    # start index = 2
    # samples = 3
    end_index = start_index + samples   # El rango es 2 hasta 4. El 5 no se incluye
                                                #cuando llega a 5 sale de la funcion. Es como less or equal to 5

    print(f"Rows from: {start_index} to {end_index - 1}") # Se pone el -1 solo para imprimir el rango correcto


    peaks_range =  df['MAXABSIMPXTE (m)'].iloc[start_index:end_index] #Del df toma la col MAXABSIMPXTE y ... 
                                                                    # los renglones 2 hasta 4. Cuando llega a 5 break.


    abs_range =  df['AVGABSIMPXTE (m)'].iloc[start_index:end_index] #Del df toma la col AVGABSIMPXTE y ... 
                                                                    # los renglones 2 hasta 4. Cuando llega a 5 break.

    #print(type(peaks_range))
    #print(abs_range)
    return peaks_range,abs_range
```

```python
ranges_peak, ranges_abs = get_ranges(df2,2,3)
```

```python
print(ranges_peak)
print(ranges_abs)
```

```python
def compute_mean(peaks_samples, abs_samples):
    avg_peaks = peaks_samples.mean()
    avg_abs = abs_samples.mean()
    return avg_abs, avg_peaks
```

```python
peak_avg,abs_avg = compute_mean(ranges_peak,ranges_abs)

print(peak_avg)
print(abs_avg)
```

```python
columns = ["PEAKS AVG", "ABS AVG"]

output_df =  pd.DataFrame(index= range(5), columns=columns)
output_df.loc[0] = [peak_avg,abs_avg]
output_df.head()
```

```python
def paste_avgs(df_results, row_in_df, temp_peak_avg,temp_abs_avg):

    df_results.iloc[row_in_df] = [temp_peak_avg,temp_abs_avg]
    print(df_results.head())
```
